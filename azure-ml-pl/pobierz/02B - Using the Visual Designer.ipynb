# Praca z Designerem wizualnym

> **Ważne**: To ćwiczenie zakłada, że masz ukończone [Lab 2A](labdocs/Lab02A.md) i [Lab 2B](labdocs/Lab02B.md) - czyli utworzony potok wnioskowania wdrożony jako punkt końcowy czasu rzeczywistego oparty na **Azure Container Instance (ACI)**.

Dodaj kod do poniższej komórki, aby przetestować swój punkt końcowy **predict-diabetes**.

W studio Azure Machine Learning otwórz punkt końcowy na stronie **Endpoints**, przejdź na kartę **Consume** i skopiuj przykładowy kod w Pythonie. Wykorzystuje on bibliotekę `requests` do wysłania żądania `POST` z ładunkiem JSON na adres URI oceny, przekazując klucz punktu końcowego w nagłówku `Authorization`. Wklej ten kod poniżej, uruchom komórkę i sprawdź, czy w odpowiedzi dostajesz przewidywaną etykietę cukrzycy oraz prawdopodobieństwo dla każdej obserwacji pacjenta.

> **Uwaga**: Klasyczny potok Designera wdraża się jako usługę sieciową w ACI (albo w AKS), a **nie** jako *zarządzany punkt końcowy online*. Oznacza to, że funkcja `ml_client.online_endpoints.invoke(...)` z SDK v2 **tutaj nie zadziała** - trzeba odwołać się bezpośrednio do adresu URI oceny przez `requests`, dokładnie tak, jak robi to przykładowy kod z karty **Consume**. Z prawdziwymi zarządzanymi punktami końcowymi online - i z `ml_client.online_endpoints.invoke(...)` - popracujesz w [Lab 7A](labdocs/Lab07A.md).
